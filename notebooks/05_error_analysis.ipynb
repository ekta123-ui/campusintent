# 05 - Error analysis

```python
import sys, os
sys.path.insert(0, os.path.abspath(".."))   # so `src` is importable from /notebooks
import pandas as pd, matplotlib.pyplot as plt
from src import config
from src.predict import IntentPredictor
from src.evaluate import most_confused
```

```python
test = pd.read_csv(config.TEST_CSV, keep_default_na=False)
p = IntentPredictor()
preds = [p.predict(q) for q in test["query"]]
test["pred"] = [r["intent"] for r in preds]; test["conf"] = [r["confidence"] for r in preds]
wrong = test[test.intent != test.pred]
print(len(wrong), "errors out of", len(test))
wrong[["query", "intent", "pred", "conf"]]
```

```python
most_confused(test['intent'], test['pred'], config.LABELS)
```

```python
# Does performance drop on noisy styles?
test["ok"] = test.intent == test.pred
test.groupby("style")["ok"].mean()
```
