# 01 - Exploratory data analysis
Class balance, query length, style and language mix.

```python
import sys, os
sys.path.insert(0, os.path.abspath(".."))   # so `src` is importable from /notebooks
import pandas as pd, matplotlib.pyplot as plt
from src import config
df = pd.read_csv(config.DATA_RAW, keep_default_na=False)
df.head()
```

```python
df.intent.value_counts().plot(kind='barh', figsize=(7,5), title='Queries per intent'); plt.show()
```

```python
df['n_words'] = df['query'].str.split().str.len()
df['n_words'].plot(kind='hist', bins=15, title='Query length (words)'); plt.show()
```

```python
print(df['style'].value_counts()); print(df['language'].value_counts()); print(df[df.course!=''].course.value_counts())
```

```python
from src.nlp_features import top_ngrams
for intent in config.LABELS:
    print(intent, top_ngrams(df[df.intent == intent]["query"], n=2, k=5))
```

```python
# Optional word cloud for one intent
from wordcloud import WordCloud
text = " ".join(df[df.intent == "Placement"]["query"])
plt.imshow(WordCloud(width=800, height=400, background_color="white").generate(text)); plt.axis("off"); plt.show()
```
