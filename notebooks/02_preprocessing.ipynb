```python
import sys, os
sys.path.insert(0, os.path.abspath(".."))   # so `src` is importable from /notebooks
import pandas as pd, matplotlib.pyplot as plt
from src import config
from src.preprocessing import clean_text, tokenize, lemmatize, preprocess
from src.nlp_features import extract_entities, pos_tags
```

```python
examples = ["Pls tell whn 3rd sem results r declared for B.Tech AI & ML?", "whats the hstl in-time for girls", "is the canteen open on Saturday?"]
for q in examples:
    print("RAW      :", q)
    print("CLEAN    :", clean_text(q))
    print("TOKENS   :", tokenize(q))
    print("LEMMAS   :", lemmatize(tokenize(q)))
    print("ENTITIES :", extract_entities(q))
    print()
```
